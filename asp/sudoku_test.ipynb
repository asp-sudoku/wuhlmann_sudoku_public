```python
%%clingo - 0

initial(1,1,1). initial(1,2,4). initial(1,3,3). %-------------% 
%-------------% initial(2,2,3). initial(2,3,1). initial(2,4,4).
initial(3,1,3). %-------------% initial(3,3,4). %-------------% 
initial(4,1,4). initial(4,2,1). %-------------% initial(4,4,3).


% set-up parameters 
size(16).
subgrid_size(2).
row(1..4).
col(1..4).
num(1..4).
{cell(X, Y) : row(X), col(Y)} = S :- size(S).

n_initials(C) :- C = #count{I, J, N : initial(I, J, N)}.
n_free(S - C) :- size(S), n_initials(C).
n_poss(4 * F) :- n_free(F).

% Generator 

% Every initial must be in the answer set.
sudoku(I, J, N) :- initial(I,J,N).

% For every non initial position, generate candiates.
P {sudoku(I, J, N) : row(I), col(J), num(N), not initial(I, J, _)} :- n_poss(P).

% Tester 

% It cannot be the case that two predicates from the same row have the same N, unless there are the same predicate.
:- sudoku(I, J, N), sudoku(I', J, N), I != I'.

% Same but for cols. 
:- sudoku(I, J, N), sudoku(I, J', N), J != J'. 

% label cells with subgrid-number 
{label(I, J, L) : sudoku(I, J, _), L = (((I-1)/S)*S + (J-1)/S), subgrid_size(S)}.

% Grid
I=I', J=J' :- sudoku(I, J, N), label(I, J, L), sudoku(I', J', N), label(I',J', L).

% Make sure, we have exactly size(N) sudoku predicates. 
:- initial(I,J,N), not sudoku(I,J,N).
:- cell(I,J), not sudoku(I,J,_).

#show sudoku/3.
```
